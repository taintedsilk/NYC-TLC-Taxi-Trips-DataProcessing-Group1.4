```python
# filename: 4_create_visualizations.py
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# --- Configuration ---
PROCESSED_DIR = Path('../processed')
FIGURES_DIR = Path('../figures')
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
sns.set_theme(style="whitegrid")

WEEKDAY_ORDER = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

def plot_monthly_trips():
    try:
        df = pd.read_csv(PROCESSED_DIR / 'kpi_monthly_2022.csv')
        # Ensure date parsing works for the index or column usually named 'date' or 'Unnamed: 0'
        if 'date' not in df.columns and 'Unnamed: 0' in df.columns:
            df.rename(columns={'Unnamed: 0': 'date'}, inplace=True)
            
        plt.figure(figsize=(10, 6))
        sns.lineplot(data=df, x='date', y='trip_count', marker='o')
        plt.title('Total Trips by Month (2022)')
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.savefig(FIGURES_DIR / 'monthly_trips_2022.png')
        plt.close()
        print("Saved Monthly Trips chart.")
    except Exception as e:
        print(f"Skipping Monthly Chart: {e}")

def plot_dow_duration():
    try:
        df = pd.read_csv(PROCESSED_DIR / 'kpi_granular_dow_duration.csv')
        
        # Sort by weekday
        df = df.set_index('Day_of_Week').reindex(WEEKDAY_ORDER).reset_index()
        
        plt.figure(figsize=(10, 5))
        sns.barplot(data=df, x='Day_of_Week', y='p95_trip_duration', palette='viridis', edgecolor='black')
        plt.title('P95 Trip Duration by Day of Week')
        plt.ylabel('Minutes')
        plt.tight_layout()
        plt.savefig(FIGURES_DIR / 'p95_duration_dow.png')
        plt.close()
        print("Saved DOW Duration chart.")
    except Exception as e:
        print(f"Skipping DOW Duration Chart: {e}")

def plot_heatmap():
    try:
        df = pd.read_csv(PROCESSED_DIR / 'kpi_granular_heatmap.csv')
        
        # Pivot for heatmap: index=Day, col=Hour, val=Count
        pivot = df.pivot(index='day_of_week', columns='hour', values='trip_count')
        pivot = pivot.reindex(WEEKDAY_ORDER)
        
        plt.figure(figsize=(14, 6))
        sns.heatmap(pivot, cmap='YlOrRd', linewidths=0.5)
        plt.title('Trip Density Heatmap: Day vs Hour')
        plt.tight_layout()
        plt.savefig(FIGURES_DIR / 'heatmap_day_hour.png')
        plt.close()
        print("Saved Heatmap.")
    except Exception as e:
        print(f"Skipping Heatmap: {e}")

def plot_speed_curve():
    try:
        df = pd.read_csv(PROCESSED_DIR / 'kpi_speed_by_hour.csv')
        
        plt.figure(figsize=(10, 5))
        sns.lineplot(data=df, x='hour_of_day', y='median_speed_mph', marker='o', color='red')
        plt.title('Median Speed by Hour of Day')
        plt.ylabel('Speed (MPH)')
        plt.xticks(range(0, 24))
        plt.grid(True, linestyle='--')
        plt.tight_layout()
        plt.savefig(FIGURES_DIR / 'speed_by_hour.png')
        plt.close()
        print("Saved Speed chart.")
    except Exception as e:
        print(f"Skipping Speed Chart: {e}")

plot_monthly_trips()
plot_dow_duration()
plot_heatmap()
plot_speed_curve()
```
