```python
# Đọc file KPI dow
file_path_kpi_dow = r'..\processed\kpi_dow_p95_trip_duration.csv'
kpi_dow_data = pd.read_csv(file_path_kpi_dow)


weekday_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Đặt lại thứ tự ngày trong tuần
kpi_dow_data = kpi_dow_data.set_index('Day_of_Week')
kpi_dow_data = kpi_dow_data.reindex(weekday_order)
kpi_dow_data = kpi_dow_data.reset_index()

# Vẽ khung
plt.figure(figsize=(10, 5))

# Biểu đồ cột biểu diễn P95 trip duration theo ngày trong tuần
sns.barplot(
    data = kpi_dow_data,
    x = 'Day_of_Week',
    y = 'p95_trip_duration',
    palette = 'viridis',
    edgecolor = 'black'
)

# Chi tiết về trục y
plt.yticks(range(0, 50, 5))

# Chú giải
plt.title('P95 Trip Duration by Day of Week in 2022', fontsize=16)
plt.xlabel('Day of Week', fontsize=12)
plt.ylabel('P95 Trip Duration (Minutes)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Lưu hình ảnh
output_path_dow = r'..\figures\p95_trip_duration_by_dow_2022.png'
plt.savefig(output_path_dow, bbox_inches='tight')

plt.show()
```
